```python
# Gerar dados clientes e produtos (limpos - tabelas de referência)
import pandas as pd
import random

random.seed(7)  # garante que os dados sejam sempre os mesmos ao rodar de novo

# Tabela de clientes
n_clientes = 300
primeiros_nomes = ["Mariana", "Rafael", "Beatriz", "Thiago", "Camila", "Lucas",
                    "Fernanda", "Gabriel", "Larissa", "André", "Juliana", "Vinícius",
                    "Patrícia", "Rodrigo", "Isabela"]
sobrenomes = ["Almeida", "Ferreira", "Cardoso", "Barbosa", "Teixeira", "Moreira",
              "Cavalcanti", "Nascimento", "Pinheiro", "Correia"]
cidades = ["Porto Alegre", "Vitória", "Natal", "Campo Grande", "João Pessoa",
           "Maceió", "Londrina", "Caxias do Sul", "Juiz de Fora", "Ribeirão Preto"]

clientes = pd.DataFrame({
    "cliente_id": range(1, n_clientes + 1),
    "nome": [f"{random.choice(primeiros_nomes)} {random.choice(sobrenomes)}" for _ in range(n_clientes)],
    "cidade": [random.choice(cidades) for _ in range(n_clientes)]
})

# Tabela de produtos - tema: papelaria e artigos criativos
n_produtos = 30
categorias_produtos = ["Cadernos", "Canetas e Lápis", "Artes e Pintura", "Organização", "Mochilas e Estojos"]
nomes_produtos = ["Caderno Universitário", "Caderno de Desenho", "Caneta Gel", "Lapiseira",
                   "Estojo Escolar", "Marcador Permanente", "Bloco de Notas Adesivo",
                   "Tinta Guache", "Pincel Aquarela", "Aquarela Profissional",
                   "Organizador de Mesa", "Pasta Catálogo", "Mochila Escolar",
                   "Estojo de Lápis de Cor", "Tesoura Escolar", "Cola Bastão",
                   "Fita Adesiva Colorida", "Régua 30cm", "Apontador com Depósito",
                   "Giz de Cera", "Papel Craft", "Álbum de Fotos", "Caneta Marca-Texto",
                   "Bloco de Rascunho", "Estojo de Canetinhas", "Squeeze Personalizado",
                   "Agenda Anual", "Post-it Colorido", "Kit Papelaria Presente", "Lápis de Cor Aquarelável"]

produtos = pd.DataFrame({
    "product_id": range(1, n_produtos + 1),
    "nome_produto": nomes_produtos,
    "categoria": [random.choice(categorias_produtos) for _ in range(n_produtos)],
    "preco": [round(random.uniform(5, 250), 2) for _ in range(n_produtos)]
})

print(clientes.head())
print(produtos.head())
```

```python
# Gerar pedidos com anomalias propositais (~15% do total)
n_pedidos = 1500
pedidos_lista = []

for pedido_id in range(1, n_pedidos + 1):
    tipo_anomalia = random.random()  # número entre 0 e 1, decide se o pedido é anômalo

    if tipo_anomalia < 0.05:
        # 5% - quantidade inválida (zero ou negativa)
        quantidade = random.choice([0, -1, -3, -8])
        cliente_id = random.randint(1, n_clientes)
        product_id = random.randint(1, n_produtos)
    elif tipo_anomalia < 0.10:
        # 5% - cliente inexistente
        quantidade = random.randint(1, 6)
        cliente_id = random.randint(n_clientes + 1, n_clientes + 500)  # fora da faixa válida
        product_id = random.randint(1, n_produtos)
    elif tipo_anomalia < 0.15:
        # 5% - produto inexistente
        quantidade = random.randint(1, 6)
        cliente_id = random.randint(1, n_clientes)
        product_id = random.randint(n_produtos + 1, n_produtos + 500)  # fora da faixa válida
    else:
        # 85% - pedido válido
        quantidade = random.randint(1, 6)
        cliente_id = random.randint(1, n_clientes)
        product_id = random.randint(1, n_produtos)

    pedidos_lista.append({
        "pedido_id": pedido_id,
        "cliente_id": cliente_id,
        "product_id": product_id,
        "quantidade": quantidade,
        "data_pedido": "2026-09-07"
    })

pedidos = pd.DataFrame(pedidos_lista)
print(pedidos.head(10))
print(f"\nTotal de pedidos: {len(pedidos)}")
```

```python
# Ler os dados de volta do S3
import io

def ler_csv_s3(bucket, key):
    obj = s3.get_object(Bucket=bucket, Key=key)
    return pd.read_csv(io.BytesIO(obj["Body"].read()))

clientes_raw = ler_csv_s3(bucket_name, f"raw/clientes/ingest_date={ingest_date}/clientes.csv")
produtos_raw = ler_csv_s3(bucket_name, f"raw/produtos/ingest_date={ingest_date}/produtos.csv")
pedidos_raw = ler_csv_s3(bucket_name, f"raw/pedidos/ingest_date={ingest_date}/pedidos.csv")

print(f"Clientes: {len(clientes_raw)} | Produtos: {len(produtos_raw)} | Pedidos: {len(pedidos_raw)}")
```

```python
# Aplicar regras de Data Quality e marcar motivo da rejeição
def identificar_motivo(row):
    motivos = []
    if row["quantidade"] <= 0:
        motivos.append("quantidade_invalida")
    if row["cliente_id"] not in clientes_raw["cliente_id"].values:
        motivos.append("cliente_inexistente")
    if row["product_id"] not in produtos_raw["product_id"].values:
        motivos.append("produto_inexistente")
    return ",".join(motivos) if motivos else None

pedidos_raw["motivo_rejeicao"] = pedidos_raw.apply(identificar_motivo, axis=1)

pedidos_validos = pedidos_raw[pedidos_raw["motivo_rejeicao"].isna()].drop(columns=["motivo_rejeicao"])
pedidos_rejeitados = pedidos_raw[pedidos_raw["motivo_rejeicao"].notna()]

print(f"Pedidos válidos: {len(pedidos_validos)}")
print(f"Pedidos rejeitados: {len(pedidos_rejeitados)}")
```

```python
# Gravar os rejeitados na Quarentena (JSON, com o motivo)
quarantine_key = f"quarantine/pedidos_rejeitados/data={ingest_date}/rejeitados.json"

json_bytes = pedidos_rejeitados.to_json(orient="records", force_ascii=False, indent=2).encode("utf-8")
s3.put_object(Bucket=bucket_name, Key=quarantine_key, Body=json_bytes)

print(f"Quarentena gravada em: s3://{bucket_name}/{quarantine_key}")
```

```python
# Construir a camada Silver: JOIN + valor_total
fato_vendas = (
    pedidos_validos
    .merge(clientes_raw, on="cliente_id", how="left")
    .merge(produtos_raw, on="product_id", how="left")
)

fato_vendas["valor_total"] = fato_vendas["quantidade"] * fato_vendas["preco"]

print(fato_vendas.head())
print(f"\nTotal de linhas na Silver: {len(fato_vendas)}")
```

```python
# Salvar a Silver em Parquet (formato colunar, mais eficiente pra análise) e subir pro S3
!pip install pyarrow -q

os.makedirs("/content/processed_local", exist_ok=True)
local_path_silver = "/content/processed_local/fato_vendas.parquet"

fato_vendas.to_parquet(local_path_silver, index=False, engine="pyarrow", compression="snappy")

silver_key = f"processed/fato_vendas/ingest_date={ingest_date}/fato_vendas.parquet"
s3.upload_file(local_path_silver, bucket_name, silver_key)

print(f"Silver gravada em: s3://{bucket_name}/{silver_key}")
```

```python
# Calcular as métricas agregadas - por categoria
gold_categoria = (
    fato_vendas
    .groupby("categoria")
    .agg(
        total_pedidos=("pedido_id", "count"),
        quantidade_total=("quantidade", "sum"),
        valor_total_vendido=("valor_total", "sum"),
        ticket_medio=("valor_total", "mean")
    )
    .reset_index()
    .round(2)
)

print(gold_categoria)
```

```python
# Calcular as métricas agregadas - por cidade
gold_cidade = (
    fato_vendas
    .groupby("cidade")
    .agg(
        total_pedidos=("pedido_id", "count"),
        quantidade_total=("quantidade", "sum"),
        valor_total_vendido=("valor_total", "sum"),
        ticket_medio=("valor_total", "mean")
    )
    .reset_index()
    .round(2)
)

print(gold_cidade)
```

```python
# Criar o cliente do Athena e definir onde ele vai guardar os resultados das queries
athena = session.client("athena")

output_location = f"s3://{bucket_name}/athena-results/"
print(f"Resultados do Athena serão salvos em: {output_location}")
```

```python
# Função auxiliar para rodar as queries e esperar o resultado
import time

def rodar_query_athena(query, database="default"):
    execution = athena.start_query_execution(
        QueryString=query,
        QueryExecutionContext={"Database": database},
        ResultConfiguration={"OutputLocation": output_location}
    )
    query_id = execution["QueryExecutionId"]

    while True:
        status = athena.get_query_execution(QueryExecutionId=query_id)
        state = status["QueryExecution"]["Status"]["State"]
        if state in ["SUCCEEDED", "FAILED", "CANCELLED"]:
            break
        time.sleep(1)

    if state != "SUCCEEDED":
        motivo = status["QueryExecution"]["Status"].get("StateChangeReason", "sem detalhes")
        raise Exception(f"Query falhou: {motivo}")

    resultado = athena.get_query_results(QueryExecutionId=query_id)
    return resultado

print("Função pronta!")
```

```python
# Criar banco de dados no Athena
db_name = f"datalake_db_{bucket_name.replace('-', '_')}"

rodar_query_athena(f"CREATE DATABASE IF NOT EXISTS {db_name}")
print(f"Banco de dados criado: {db_name}")
```

```python
# Criar as tabelas externas da camada raw (CSV)
query_raw_clientes = f"""
CREATE EXTERNAL TABLE IF NOT EXISTS {db_name}.raw_clientes (
    cliente_id INT,
    nome STRING,
    cidade STRING
)
PARTITIONED BY (ingest_date STRING)
ROW FORMAT DELIMITED
FIELDS TERMINATED BY ','
LOCATION 's3://{bucket_name}/raw/clientes/'
TBLPROPERTIES ('skip.header.line.count'='1')
"""
rodar_query_athena(query_raw_clientes, database=db_name)
print("Tabela raw_clientes criada")

query_raw_produtos = f"""
CREATE EXTERNAL TABLE IF NOT EXISTS {db_name}.raw_produtos (
    product_id INT,
    nome_produto STRING,
    categoria STRING,
    preco DOUBLE
)
PARTITIONED BY (ingest_date STRING)
ROW FORMAT DELIMITED
FIELDS TERMINATED BY ','
LOCATION 's3://{bucket_name}/raw/produtos/'
TBLPROPERTIES ('skip.header.line.count'='1')
"""
rodar_query_athena(query_raw_produtos, database=db_name)
print("Tabela raw_produtos criada")

query_raw_pedidos = f"""
CREATE EXTERNAL TABLE IF NOT EXISTS {db_name}.raw_pedidos (
    pedido_id INT,
    cliente_id INT,
    product_id INT,
    quantidade INT,
    data_pedido STRING
)
PARTITIONED BY (ingest_date STRING)
ROW FORMAT DELIMITED
FIELDS TERMINATED BY ','
LOCATION 's3://{bucket_name}/raw/pedidos/'
TBLPROPERTIES ('skip.header.line.count'='1')
"""
rodar_query_athena(query_raw_pedidos, database=db_name)
print("Tabela raw_pedidos criada")
```
